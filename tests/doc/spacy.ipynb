```python
import spacy
from spacy import displacy
import cairosvg
nlp = spacy.load("en_core_web_sm")

```

```python
text = "Test the software, if it has not been tested. Do the test one more time, when the result is incorrect."

doc = nlp(text)
displacy.render(doc, style="dep", jupyter=True)
svg = displacy.render(doc, style="dep", jupyter=False)
cairosvg.svg2png(bytestring=svg.encode("utf-8"), write_to="displacy1.png")
```
